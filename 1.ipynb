```python
import matplotlib.pyplot as plt
import numpy as np

# Input range
x = np.linspace(-10, 10, 10)

# Activation functions
def sigmoid(x):
    return 1 / (1 + np.exp(-x))

def tanh_fn(x):
    return np.tanh(x)

def relu(x):
    return np.maximum(0, x)

# Plot
plt.figure(figsize=(10, 10))
plt.suptitle("Activation Functions", color="blue", fontsize=16)

# 1. Sigmoid
plt.subplot(2, 2, 1)
plt.plot(x, sigmoid(x), label="Sigmoid")
plt.scatter(x, sigmoid(x), s=8, color="black")
plt.xlabel("Input")
plt.ylabel("Output")
plt.legend()
plt.grid(True)

# 2. Tanh
plt.subplot(2, 2, 2)
plt.plot(x, tanh_fn(x), label="Tanh", color="green")
plt.scatter(x, tanh_fn(x), s=8, color="black")
plt.xlabel("Input")
plt.ylabel("Output")
plt.legend()
plt.grid(True)

# 3. ReLU
plt.subplot(2, 2, 3)
plt.plot(x, relu(x), label="ReLU", color="brown")
plt.scatter(x, relu(x), s=8, color="black")
plt.xlabel("Input")
plt.ylabel("Output")
plt.legend()
plt.grid(True)

plt.show()
```

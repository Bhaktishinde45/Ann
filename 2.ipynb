```python
import numpy as np

def ANDNOT():
    w1 = 1
    w2 = -1

    # Input and expected output
    x = np.array([[1,1], [1,0], [0,1], [0,0]])
    y = [0, 1, 0, 0]

    t = int(input("Enter Threshold: "))

    print("\nx1  x2  Output")
    outputs = []

    for xi in x:
        net = w1 * xi[0] + w2 * xi[1]
        out = 1 if net >= t else 0
        outputs.append(out)
        print(xi[0], " ", xi[1], " ", out)

    # Check correctness
    if outputs == y:
        print("\nCorrect threshold!")
    else:
        print("\nWrong threshold. Try again.\n")
        ANDNOT()

ANDNOT()
```

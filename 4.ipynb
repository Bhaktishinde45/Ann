```python
import numpy as np
import matplotlib.pyplot as plt

# Data
X = np.array([[0,0],[0,1],[1,0],[1,1]])
y = np.array([0,0,0,1])

# Init
w = np.zeros(2); b = 0; lr = 0.1

# Training
for e in range(50):
    for xi, yi in zip(X, y):
        pred = 1 if np.dot(xi, w) + b >= 0 else 0
        err = yi - pred
        w += lr * err * xi; b += lr * err

print("Weights:", w, "Bias:", b)

# Testing
print("\nResults:")
for xi, yi in zip(X, y):
    print(xi, "->", 1 if np.dot(xi, w) + b >= 0 else 0)

# Plot
plt.scatter(X[:,0], X[:,1], c=y, cmap='bwr', s=120)

x = np.linspace(-0.2, 1.2, 50)
if w[1] != 0:
    plt.plot(x, -(w[0]*x + b)/w[1])
else:
    plt.axvline(x=-b/w[0])

plt.title("AND Gate")
plt.xlabel("x1"); plt.ylabel("x2")
plt.grid(); 
plt.show()
```

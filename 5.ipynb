```python

import numpy as np 
x1=np.array([-1,1,1]) 
x2=np.array([1,-1,-1]) 
y1=np.array([1,1]) 
y2=np.array([-1,-1]) 
print("************Initially***************") 
print("x1:",x1) 
print("y1:",y1) 
print("-----------") 
print("x2:",x2) 
print("y2:",y2) 
w=np.outer(x1,y1)+np.outer(x2,y2) 
print("Weight matrix:") 
print(w) 
print("************FORWARD***************") 
y_1=np.sign(np.dot(x1,w)) 
print("input:","x1:",x1) 
print("output:","y1:",y_1) 
y_2=np.sign(np.dot(x2,w)) 
print("input:","x2:",x2) 
print("output:","y2:",y_2)  
print("************BACKWARD*************") 
x_1=np.sign(np.dot(w,y1)) 
print("input:","y1:",y1) 
print("output:","x1:",x_1) 
x_2=np.sign(np.dot(w,y2)) 
print("input:","y2:",y2) 
print("output:","x2:",x_2)
```

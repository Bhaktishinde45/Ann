```python
import numpy as np

X = np.array([[0,0],[0,1],[1,0],[1,1]])
Y = np.array([[0],[1],[1],[0]])

np.random.seed(0)
W1 = np.random.rand(2,2)
W2 = np.random.rand(2,1)

def s(x):
    return 1/(1+np.exp(-x))

def sd(x):
    return x*(1-x)

lr = 0.1
for i in range(5000):
    H = s(np.dot(X, W1))
    O = s(np.dot(H, W2))
    e = Y - O
    dO = e * sd(O)
    dH = np.dot(dO, W2.T) * sd(H)
    W2 += np.dot(H.T, dO) * lr
    W1 += np.dot(X.T, dH) * lr

print("Final W1:", W1)
print("Final W2:", W2)
print("Output:", O)

FO = [1 if v > 0.5 else 0 for v in O]
print("Final Output:", FO)

for i in range(4):
    print(X[i], "|", Y[i], "|", FO[i])
```
